## 6. Your individual project challenge
**Challenge:** Calculate revenue; identify best-selling product and category-wise revenue; rank products.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
df["revenue"]=df["Unit_Price"]*df["Quantity"]
best_selling_product = df.groupby("Product")["Quantity"].sum().idxmax()
print("Best selling product",best_selling_product)

category_revenue=df.groupby("Category")["revenue"].sum().sort_values(ascending=False)
print("\ncategory-wise Revenue :\n",category_revenue)
product_ranking=( df.groupby("Product")["revenue"].sum().sort_values(ascending=False).reset_index())
product_ranking["rank"]=product_ranking["revenue"].rank(ascending=False,method="dense").astype(int)
print("\nProduct Ranking by Revenue:\n",product_ranking)
```
